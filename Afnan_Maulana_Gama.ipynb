```python
panjang_persegi_panjang = int(input("masukan nilai panjang dari persegi panjang (cm)"))
lebar_persegi_panjang = int(input("masukan nilai lebar dari persegi panjang (cm)"))

luas_persegi_panjang = panjang_persegi_panjang * lebar_persegi_panjang

print(f"Luas persegi panjang yaitu {luas_persegi_panjang} cm^2")
```

```python
panjang_balok = float(input("Masukan nilai panjang dari balok (cm): "))
lebar_balok = int(input("Masukan nilai lebar dari balok (cm): "))
tinggi_balok = float(input("Masukan nilai tinggi dari balok (cm): "))

volume_balok = panjang_balok * lebar_balok * tinggi_balok

print(f"Volume balok yaitu: {volume_balok} cm^3")
```
